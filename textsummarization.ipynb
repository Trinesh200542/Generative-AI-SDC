```python
# Step 1: Install transformers library
!pip install transformers

# Step 2: Import required modules
from transformers import pipeline

# Step 3: Load the summarization pipeline
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# Step 4: Provide your input text
text = """
Climate change refers to long-term shifts in temperatures and weather patterns. These shifts may be natural, such as through variations in the solar cycle. But since the 1800s, human activities have been the main driver of climate change, primarily due to burning fossil fuels like coal, oil, and gas.

Burning these materials releases what are called greenhouse gases into Earth’s atmosphere. These emissions trap heat from the sun, causing the planet’s temperature to rise. This phenomenon is known as global warming, and it is just one aspect of climate change.

As the planet warms, we experience more extreme weather events—like heatwaves, droughts, and heavy rainfall. Sea levels are rising as glaciers and ice sheets melt, threatening coastal communities. Wildlife and ecosystems are being disrupted, with many species struggling to adapt to rapidly changing environments.

Climate change is not a distant threat—it’s already affecting people around the world. From more intense hurricanes in the Atlantic, to devastating wildfires in Australia and California, the signs are becoming increasingly hard to ignore. In many parts of the world, agricultural productivity is declining due to shifts in rainfall and temperature, putting food security at risk.

The good news is, we can still limit the worst effects of climate change if we act quickly. Solutions include transitioning to renewable energy sources like wind and solar, improving energy efficiency, and protecting forests which absorb carbon dioxide. On a personal level, people can help by reducing car travel, conserving energy, and supporting sustainable practices.

Governments, businesses, and individuals all have a role to play. International agreements like the Paris Accord aim to bring countries together to set emissions targets and promote climate action. But progress has been slow, and urgent, large-scale changes are still needed.

Ultimately, addressing climate change requires a collective effort. By working together, we can protect the planet for future generations and build a more sustainable and equitable world.
"""

# Step 5: Generate two different summaries
summary1 = summarizer(text, max_length=60, min_length=25, do_sample=True, top_k=50)[0]['summary_text']
summary2 = summarizer(text, max_length=40, min_length=15, do_sample=True, top_p=0.92)[0]['summary_text']

# Step 6: Print the results
print("Original Text:\n", text)
print("\n🔹 Summarized Text 1:\n", summary1)
print("\n🔹 Summarized Text 2:\n", summary2)
```
